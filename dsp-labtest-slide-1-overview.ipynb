# Plotting Cosine Graph

```python
time = np.arange(0, 10, 0.1)
amplitude = np.cos(time)

plt.plot(time, amplitude)
plt.title("Cosine Wave (Amp Vs. Time)")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True, which="both")
plt.axhline(y=0, color="k")
plt.axvline(x=0, color="k")
```

# Plotting Exponential Graph (e^-t)

```python
time = np.arange(0, 10, 0.1)
amplitude = np.exp(-time)

plt.plot(time, amplitude)
plt.title("Exponential Signal (Amplitude vs Time)")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True, which="both")
plt.axhline(y=0, color="k")
plt.axvline(x=0, color="k")
plt.show()
```

# Plotting Graph for Unit Step Signal

```python
time = np.arange(-5, 5, 0.1)
amplitude = np.where(np.sin(time) >= 0, 1, 0) # If time >= 0, put 1; otherwise put 0.

plt.plot(time, amplitude)
plt.title("Unit Step Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True, which="both")
plt.axhline(y=0, color="k")
plt.axvline(x=0, color="k")
plt.show()
```

# Plotting Graph for Ramp Signal

```python
time = np.arange(0, 5, 0.1)
amplitude = time % 1  #everytime it reaches 1, it starts over

plt.plot(time, amplitude)
plt.title("Ramp Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True, which="both")
plt.axhline(y=0, color="k")
plt.axvline(x=0, color="k")
plt.show()
```

# Plotting Graph for Square Wave

```python
time = np.arange(0, 10, 0.1)
amplitude = np.sign(np.sin(time))

plt.plot(time, amplitude)
plt.title("Square Wave")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True, which="both")
plt.axhline(y=0, color="k")
plt.axvline(x=0, color="k")
plt.show()
```
