# Plotting the Graphs
# (Original, Delay, Advance, Reflection, Upsampling, Downsampling)

```python
import numpy as np
import matplotlib.pyplot as plt


#ORIGINAL SIGNAL, x[n]
signal = [-1,2,-4,5,6,4,5,5,7,8,-4,-6,-5,-4,-3,-2] # amplitudes (16 samples)
origin = 5 #(6th sample, index 5, value = 4)

time = np.arange(-origin, len(signal) - origin)

plt.figure(figsize=(12, 15))
plt.subplot(3,2,1)
plt.stem(time, signal, "r+")
plt.title("Original Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True, which="both")



#DELAYED SIGNAL, x[n-k]
delay = 4
time_delay = np.arange(-origin + delay, len(signal) - origin + delay)


plt.subplot(3,2,2) #Plots the graph in the 2nd box
plt.stem(time_delay, signal, "r+")
plt.title("Delayed Signal (x[n-4])")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
#plt.grid(True, which="both")



#ADVANCED SIGNAL, x[n+k]
advance = 4
time_advance = np.arange(-origin - advance, len(signal) - origin - advance)


plt.subplot(3,2,3) #Plots the graph in the 3rd box
plt.stem(time_advance, signal, "r+")
plt.title("Advanced Signal (x[n+4])")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True, which="both")



#REFLECTED SIGNAL, x[-n]
reflected_time = time

plt.subplot(3,2,4) #Plots the graph in the 4th box
plt.stem(reflected_time, signal, "r+")
plt.title("Reflected Signal (x[-n])")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True, which="both")




#UPSAMPLING / Stretching of x[n] signal x[n/4]
scale = 4

time_scale = np.linspace(
    -origin * scale,
    (len(signal) - origin) * scale,
    len(signal)
) 
# time_scale = np.arange(-origin, len(signal) - origin) * scale 
#np.arange() can be used too

#np.linspace(start,end,number_of_samples)

plt.subplot(3,2,5)
plt.stem(time_scale, signal,'g+')
plt.title("Scaled Signal (Stretched) : x[n/4]")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True, which="both")




#DOWNSAMPLING / Compression of x[n] signal x[4n]
scale = 0.25

time_scale = np.arange(-origin, len(signal) - origin) * scale

plt.subplot(3,2,6)
plt.stem(time_scale, signal,'g+')
plt.title("Scaled Signal (Compression) : x[4n]")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True, which="both")
```

# Amplitude Scaling (Change how tall the Signal is)

# Code-Theory

```python
signal = np.array([-1,2,-4,5,6,4,5,5,7,8,-4,-6,-5,-4,-3,-2])
#took array cause, we want to multiply the entire signal by a number.

scaling_factors = [2,0.5,-1.5]
#These are the factors that will multiply the signal.
#2 doubles the Amp.
#0.5 makes it half.
#-1.5 flips and enlarges it.

for i, scale in enumerate(scaling_factors, 0):
    scaled_signal = scale + signal

    i = i + 1

    plt.subplot(2,2,i+1)

    plt.stem(scaled_signal)

#enumerate tells us, which Number-th iteration is this.
#just writing (for scale in scaling_factors) would not give us the Iteration-th number

```

# Amplitude Scaling (Code)

```python
signal = np.array([-1,2,-4,5,6,4,5,5,7,8,-4,-6,-5,-4,-3,-2])
scaling_factors = [2,0.5,-1.5]

plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
plt.stem(signal)
plt.title("Original Signal")
plt.xlabel("Time (n)")
plt.ylabel("Amplitude")
plt.grid(True, which="both")

for i, scale in enumerate(scaling_factors, 0):
    scaled_signal = scale * signal

    i = i + 1

    plt.subplot(2,2,i+1)
    plt.stem(scaled_signal)
    plt.title(f"Scaled Signal (x[n] : {scale})")
    plt.xlabel("Time (n)")
    plt.ylabel("Amplitude")
    plt.grid(True, which="both")
    
plt.tight_layout() #eliminates overlaps
```
