## Study and Implementation of Z-Transformation Using Python

### (The Z-transform converts a discrete-time signal into its Z-domain (complex frequency-domain) representation for easier analysis)

```python
import sympy as sp #performs algebra, calculus, summation, symbolic math
import numpy as np
import matplotlib.pyplot as plt
```

## Z-Transform of Unit Step Signal

### Given Signal

$$
x[n]=u[n]
$$

### General Z-Transform Formula

$$
X(z)=\sum_{n=0}^{\infty}x[n]z^{-n}
$$

### Substitute \(x[n]=1\)

$$
X(z)=\sum_{n=0}^{\infty}z^{-n}
$$

### Expected Z-Transform

$$
X(z)=\frac{z}{z-1}
$$

```python
n, z = sp.symbols('n z')   #symbols() -> Treat these as mathematical variables
 
X = sp.summation(z**(-n), (n, 0, sp.oo)) #(variable,start,end)


sp.pprint(X)  #just print()'s output doesn't look good
```

## Z-Transform of Exponential Signal

Given Signal:

$$
x[n]=a^n u[n]
$$

General Z-Transform Formula:

$$
X(z)=\sum_{n=0}^{\infty}x[n]z^{-n}
$$

Substitute \(x[n]=a^n\):

$$
X(z)=\sum_{n=0}^{\infty}a^nz^{-n}
$$

Expected Z-Transform:

$$
X(z)=\frac{z}{z-a}
$$

```python
n, z, a = sp.symbols('n z a')

X = sp.summation((a**n) * z**(-n), (n, 0, sp.oo))

sp.pprint(X)
```

## Z-Transform of Ramp Signal

### Given Signal

$$
x[n]=nu[n]
$$

### General Z-Transform Formula

$$
X(z)=\sum_{n=0}^{\infty}x[n]z^{-n}
$$

### Substitute \(x[n]=n\)

$$
X(z)=\sum_{n=0}^{\infty}nz^{-n}
$$

### Expected Z-Transform

$$
X(z)=\frac{z}{(z-1)^2}
$$

```python
n, z = sp.symbols('n z')

X = sp.summation(n * z**(-n), (n, 0, sp.oo))

sp.pprint(X)
```

## Inverse Z-Transform

## Z-Transform of Delayed Impulse Signal

### Given Signal

$$
x[n]=\delta[n-k]
$$

### General Z-Transform Formula

$$
X(z)=\sum_{n=0}^{\infty}x[n]z^{-n}
$$

### Substitute \((x[n]=\delta[n-k])\)

$$
X(z)=\sum_{n=0}^{\infty}\delta[n-k]z^{-n}
$$

### Expected Z-Transform

$$
X(z)=z^{-k}
$$

```python
n, z, k = sp.symbols('n z k', integer=True)

X = sp.summation(sp.KroneckerDelta(n, k) * z**(-n), (n, 0, sp.oo))

sp.pprint(X)
```
